### 9.17利用高斯金字塔对图像进行下采样

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt


def pyramid(image_rgb,level = 5):
    '''
    size must be 2**n
    '''
    rows, cols ,channesl = image_rgb.shape
    assert rows > 2**level 
    assert cols > 2**level
    image = image_rgb.copy()
    pyramid_list =[]
    pyramid_list.append(image)
    for i in range(1,level+1):        
        i_level_image = cv2.pyrDown(image)
        pyramid_list.append(i_level_image)
        image = i_level_image.copy()
    
    return pyramid_list





image = cv2.imread('campus_1024_768.jpg')
b,g,r = cv2.split(image)
image_rgb = cv2.merge([r,g,b])
pyramid_list = pyramid(image_rgb)


for i,img in enumerate(pyramid_list):
    plt.figure(figsize = (10/2**i,10/2**i))
    plt.imshow(pyramid_list[i])
    plt.show()


def getPyramidImage(pyramid_list):
    image = pyramid_list[0]
    r,c,channels = image.shape
    py_image = np.zeros((r,int(c+c/2),channels),dtype = np.uint8)
    py_image[0:r,0:c,:]= pyramid_list[0]
    row_start = 0 
    row_end = int(r/2)
    col_start = c
    col_end = int(c+c/2)
    for i  in range(1,len(pyramid_list)):   
        cur_img = pyramid_list[i]
        print(i,cur_img.shape )
        print(row_start,row_end,col_start,col_end)
        rows,cols ,channels = cur_img.shape
        py_image[row_start:row_end,col_start:col_end,:] = cur_img[:,:,:]
        row_start = row_end 
        row_end = int(row_start+rows/2)
        col_start = c
        col_end = int(c + cols/2)
    return py_image



   

plt.figure(figsize = (10,15))
plt.imshow(getPyramidImage(pyramid_list))
plt.show()



```

### 9.18利用拉普拉斯金字塔对图像进行上采样

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt


def getGaussPyramidList(image_rgb,level = 5):
    rows, cols ,channesl = image_rgb.shape
    assert rows > 2**level 
    assert cols > 2**level
    image = image_rgb.copy()
    pyramid_list =[]
    pyramid_list.append(image)
    for i in range(1,level+1):        
        i_level_image = cv2.pyrDown(image)
        pyramid_list.append(i_level_image)
        image = i_level_image.copy()
    
    return pyramid_list

def getPyramidImage(pyramid_list):
    image = pyramid_list[0]
    r,c,channels = image.shape
    py_image = np.zeros((r,int(c+c/2),channels),dtype = np.uint8)
    py_image[0:r,0:c,:]= pyramid_list[0]
    row_start = 0 
    row_end = int(r/2)
    col_start = c
    col_end = int(c+c/2)
    for i  in range(1,len(pyramid_list)):   
        cur_img = pyramid_list[i]
        print(i,cur_img.shape )
        print(row_start,row_end,col_start,col_end)
        rows,cols ,channels = cur_img.shape
        py_image[row_start:row_end,col_start:col_end,:] = cur_img[:,:,:]
        row_start = row_end 
        row_end = int(row_start+rows/2)
        col_start = c
        col_end = int(c + cols/2)
    return py_image


def getLapPyrList(image_rgb):
    pyramid_list  = getGaussPyramidList(image_rgb)
    lap_pyr_list = []
    levels = len(pyramid_list)-1
    for i in range(levels,0,-1):
        cur_gau_img= pyramid_list[i]
        up_img = cv2.pyrUp(cur_gau_img)
        lap_img = cv2.subtract(pyramid_list[i-1],up_img)
        lap_pyr_list.append(lap_img)
    lap_pyr_list = lap_pyr_list[::-1].copy()  
    return lap_pyr_list


image = cv2.imread('campus_1024_768.jpg')
b,g,r = cv2.split(image)
image_rgb = cv2.merge([r,g,b])

pyramid_list = getGaussPyramidList(image_rgb)
lap_pyr_list = getLapPyrList(image_rgb)
cur_img = pyramid_list[5]
print(cur_img.shape)
image2X = cv2.pyrUp(cur_img)
plt.imshow(image2X) 
plt.show()


plt.imshow(getPyramidImage(lap_pyr_list)*2)
plt.show()
for i,img in enumerate(lap_pyr_list):
    plt.figure(figsize = (10/2**i,10/2**i))
    plt.imshow(lap_pyr_list[i]*2)
    plt.show()










```

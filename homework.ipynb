```python
import cv2 

import numpy as np 
import matplotlib.pyplot as plt
def gaussian(x,sigma, mean=0):
    f = 1/(np.sqrt(2*np.pi)*sigma)*np.exp(-x*x/(2*sigma*sigma))
    return f
```

```python
x = np.arange(-3,4)
ksize = 7
nx = ksize-1
sigma = (nx-1)/2.0*0.3 + 0.8
print(x)
print(sigma)

mean = 0
g = gaussian(x,sigma,mean)
g2d = g.reshape(7,1) @ g.reshape(1,7)
print(g)
print(g.shape)
print(np.round(g2d/np.sum(g2d),decimals= 2))

np.sum(g2d)


```

```python
import numpy as np 
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm


def f_log(x,y,sigma):
    result = ((x*x + y*y -2*sigma*sigma)/(sigma*sigma*sigma*sigma)) * np.exp(-(x*x+y*y)/(2*sigma*sigma))
    return result

def LoG(ksize=3):
    nx = ksize -1 
    sigma = (nx-1)/2*0.3 + 0.8
    d = nx//2
    x = np.arange(-d,d+1)
    y = np.arange(-d,d+1)
    result = np.zeros((ksize,ksize),dtype = np.float64)
    for i in x:
        for j in y:
            #print(i,j)
            result[i+d,j+d] = f_log(i,j,sigma)    
    return result

def plotLoG(ksize = 3):
    nx = ksize -1 
    sigma = (nx-1)/2*0.3 + 0.8
    d = nx//2
    x = np.arange(-d,d+1)
    y = np.arange(-d,d+1)
    xx,yy = np.meshgrid(x,y)
    zz = LoG(ksize)
    fig = plt.figure()
    ax = Axes3D(fig) # 在窗口上添加3D坐标轴
    fig.add_axes(ax)#使用 ax = Axes3D(fig) ，将自身添加到图像中
    surf = ax.plot_surface(xx,yy,zz,cmap=cm.gist_rainbow)
    fig.colorbar(surf, shrink=0.5, aspect=5)
    plt.show()


print(np.around(LoG(71)))
plotLoG(7)
plotLoG(15)
plotLoG(31)
plotLoG(71)

print(np.around(LoG(3),decimals=2))
print(np.around(LoG(7),decimals=2))
print(np.around(LoG(11),decimals=2))
print(np.around(LoG(71),decimals=2))
plt.imshow(LoG(11),cmap='gray')
plt.show()


plt.imshow(LoG(71),cmap='gray')
plt.show()




```
